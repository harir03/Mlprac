```python
# ------- Step 1: Import required libraries -------
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# >>> CHANGE THIS DATA (your own readings/values) and re-run: Runtime -> Run all <<<
# ------- Step 2: Create the engineering datasets -------
# SIMULATED SAMPLE DATA: generated in this cell for demonstration,
# NOT real measured readings.
np.random.seed(42)  # fixed seed so the results are reproducible

# Stress-strain data (tensile test)
strain = np.array([0.0, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0])  # in %
stress = np.array([0, 50, 100, 148, 190, 210, 225, 230, 228])    # in MPa

# Temperature of a machine over 24 hours
time_hr = np.arange(0, 24, 1)
temperature = 25 + 10 * np.sin(time_hr / 24 * 2 * np.pi) + np.random.normal(0, 1, 24)

# Sensor measurement errors (for the histogram)
measurement_errors = np.random.normal(0, 0.5, 200)

# Material strength comparison (for the bar graph)
materials = ['Steel', 'Aluminum', 'Copper', 'Titanium', 'Brass']
tensile_strength = [400, 310, 220, 900, 350]  # in MPa

print("Peak temperature : %.2f degC at t = %d h"
      % (temperature.max(), time_hr[temperature.argmax()]))
print("Mean error : %.3f Std of error: %.3f"
      % (measurement_errors.mean(), measurement_errors.std()))
print("Strongest material:",
      materials[int(np.argmax(tensile_strength))],
      "-", max(tensile_strength), "MPa")
```

```python
# ------- Step 3: Line graph - temperature vs time -------
plt.figure(figsize=(8, 5))
plt.plot(time_hr, temperature, color='blue', marker='o', linestyle='-')
plt.title("Machine Temperature Variation Over 24 Hours")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (degC)")
plt.grid(True)
plt.show()
```

```python
# ------- Step 4: Scatter plot - stress vs strain -------
plt.figure(figsize=(8, 5))
plt.scatter(strain, stress, color='red', marker='x', s=80)
plt.title("Stress vs Strain (Tensile Test)")
plt.xlabel("Strain (%)")
plt.ylabel("Stress (MPa)")
plt.grid(True)
plt.show()
```

```python
# ------- Step 5: Histogram - sensor measurement errors -------
plt.figure(figsize=(8, 5))
plt.hist(measurement_errors, bins=20, color='green', edgecolor='black')
plt.title("Distribution of Sensor Measurement Errors")
plt.xlabel("Error (units)")
plt.ylabel("Frequency")
plt.grid(axis='y', alpha=0.7)
plt.show()
```

```python
# ------- Step 6: Bar graph - tensile strength of materials -------
plt.figure(figsize=(8, 5))
plt.bar(materials, tensile_strength,
        color=['gray', 'silver', 'orange', 'blue', 'goldenrod'])
plt.title("Tensile Strength Comparison of Engineering Materials")
plt.xlabel("Material")
plt.ylabel("Tensile Strength (MPa)")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
# ------- Step 7: Combining multiple plots (subplots) -------
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
axs[0, 0].plot(time_hr, temperature, color='blue')
axs[0, 0].set_title("Temperature vs Time")
axs[0, 1].scatter(strain, stress, color='red')
axs[0, 1].set_title("Stress vs Strain")
axs[1, 0].hist(measurement_errors, bins=20, color='green')
axs[1, 0].set_title("Measurement Error Distribution")
axs[1, 1].bar(materials, tensile_strength, color='orange')
axs[1, 1].set_title("Material Strength Comparison")
plt.tight_layout()
plt.show()
```

```python
# ------- Step 8: Customizing plots -------
peak_idx = np.argmax(temperature)  # position of the highest temperature
plt.figure(figsize=(8, 5))
plt.plot(time_hr, temperature, label="Temp (degC)", color='purple',
         linewidth=2, linestyle='--')
plt.legend()
plt.title("Customized Line Plot Example")
plt.xlabel("Time (hr)")
plt.ylabel("Temperature (degC)")
plt.annotate("Peak",
             xy=(time_hr[peak_idx], temperature[peak_idx]),
             xytext=(time_hr[peak_idx] + 4, temperature[peak_idx] - 0.5),
             arrowprops=dict(facecolor='black', shrink=0.05))
plt.show()
```
